```python
# Maior valor:
my_list = [17, 3, 11, 5, 1, 9, 7, 15, 13]
largest = my_list[0]

for i in range(1, len(my_list)):
    if my_list[i] > largest:
        largest = my_list[i]

print(largest)
```

```python
my_list = [17, 3, 11, 5, 1, 9, 7, 15, 13]
largest = my_list[0]
 
for i in my_list:
    if i > largest:
        largest = i
 
print(largest)
```

```python
my_list = [1, 2, 4, 4, 1, 4, 2, 6, 2, 9]
sem_duplicatas = []

for num in my_list:
    if num not in sem_duplicatas:
        sem_duplicatas.append(num)

print("A lista com os elementos exclusivos aqui")
print(sem_duplicatas)
```

```python
# list comprehension

squares = [x ** 2 for x in range(10)]
odds = [x for x in squares if x % 2 != 0 ]

print(squares)
print(odds)
```

```python
EMPTY = "."
ROOK = "R"
KNIGHT = "N"
PAWN = "P"
board = [[EMPTY for i in range(8)] for j in range(8)]
board[0][0] = ROOK
board[0][7] = ROOK
board[7][0] = ROOK
board[7][7] = ROOK
 
board[4][2] = KNIGHT
 
board[3][4] = PAWN
 
def imprimir_tabuleiro(tabuleiro):
    print("   0 1 2 3 4 5 6 7")  
    print("  -----------------")
    for indice_linha, linha in enumerate(tabuleiro):
        print(f"{indice_linha} | {' '.join(linha)} |")
    print("  -----------------")

# Executa a função
imprimir_tabuleiro(board)
```
