```python
# Nova lista 
my_list = [] # Criando uma lista vazia.

for i in range(5):
   my_list.append(i + 1)

print(f"\nNova lista preenchida com append e range(5): {my_list}")

my_list = []  # lista vazia
for i in range(5):
    my_list.insert(0, i + 1)
 
print(f"\nNova lista invertida preenchida com insert e range(5): {my_list}") 
```

Variação do loop *for* para processar listas

```python
my_list = [10, 1, 8, 3, 5]
total = 0

for i in range(len(my_list)):
  total += my_list[i]
print(total)
```

```python
my_list = [10, 1, 8, 3, 5]
total = 0
 
for i in my_list:
    total += i
print(total)
```

```python
variable_1 = 1
variable_2 = 2
 
auxiliary = variable_1
variable_1 = variable_2
variable_2 = auxiliary

# é a mesma coisa que:
 
variable_1 = 1
variable_2 = 2
 
variable_1, variable_2 = variable_2, variable_1
```

```python
my_list = [10, 1, 8, 3, 5]
 
my_list[0], my_list[4] = my_list[4], my_list[0]
my_list[1], my_list[3] = my_list[3], my_list[1]
 
print(my_list)
```

```python
length = len(my_list)
for i in range(length // 2):
    my_list[i], my_list[length - i - 1] = my_list[length - i - 1], my_list[i]
 
print(my_list)
 
```
