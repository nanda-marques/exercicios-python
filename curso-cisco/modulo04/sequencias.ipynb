Um tipo de sequência é um tipo de dados em Python que é capaz de armazenar mais de um valor (ou menos de um, pois uma sequência pode estar vazia), e esses valores podem ser sequencialmente (por isso o nome) navegados, elemento a elemento.

Como o loop for é uma ferramenta especialmente projetada para iterar através de sequências, podemos expressar a definição como: uma sequência são dados que podem ser verificados pelo for loop.

A segunda noção - *mutabilidade* - é uma propriedade de qualquer dado Python que descreva sua disponibilidade para ser livremente alterado durante a execução do programa. 

Os dados mutáveis podem ser atualizados livremente a qualquer momento - chamamos isso de operação in situ (em posição). Ex: *list.append(1)*

Os dados imutáveis podem se comportar como uma lista, mas não podem ser modificado in situ.

```python
# Tuplas

tuple_1 = (1, 2, 4, 8) # ou vazia com ()
tuple_2 = 1., .5, .25, .125
```

O dicionário é outra estrutura de dados Python. Não é um tipo de sequência (mas pode ser facilmente adaptado ao processamento de sequência) e é mutável.

```python
dictionary = {"gato": "chat", "cachorro": "chien", "cavalo": "cheval"}
phone_numbers = {'fefe': '11 12345-6789', 'lala': '11 1234-1111'}
empty_dictionary = {}

print(dictionary)
print(phone_numbers)
print(empty_dictionary)
```

```python
# a palavra que você procura é chamada de key. A palavra que você recebe do dicionário é chamada de value
print(dictionary['gato'])
print(phone_numbers['fefe'])
```

```python
dictionary = {"gato": "chat", "cachorro": "chien", "cavalo": "cheval"}
 
for key in dictionary.keys():
    print(key, "->", dictionary[key])
```

```python
dictionary = {"gato": "chat", "cachorro": "chien", "cachorro": "cheval"}
 
dictionary['gato'] = 'cat'
dictionary['cachorro'] = 'dog'
dictionary['cavalo'] = 'horse'
print(dictionary)
```

```python
for key in sorted(dictionary.keys()):
    print(key, "->", dictionary[key])

for english in dictionary.values():
    print(english)
```

```python
dictionary['peixe'] = 'fish'
print(dictionary)

dictionary.update({"pato": "duck"})
print(dictionary)
```

```python
del dictionary['cachorro']
print(dictionary)

dictionary.popitem()
print(dictionary)
```

```python
# Dicionários e tuplas

school_class = {}

while True:
 name = input("Digite o nome do aluno: ")
 if name == '':
    break
 
 score = int(input("Insira a pontuação do aluno (0-10): "))
 if score not in range(0, 11):
    break
 
 if name in school_class:
    school_class[name] += (score,)
 else:
    school_class[name] = (score,)
 
for name in sorted(school_class.keys()):
 adding = 0
 counter = 0
 for score in school_class[name]:
    adding += score
    counter += 1
 print(name, ":", adding / counter)
```

```python
# Exemplo 1
tuple_1 = (1, 2, 3)
for elem in tuple_1:
    print(elem)
 
# Exemplo 2
tuple_2 = (1, 2, 3, 4)
print(5 in tuple_2)
print(5 not in tuple_2)
 
# Exemplo 3
tuple_3 = (1, 2, 3, 4)
print(len(tuple_3))
print(5 not in tuple_3)

# Exemplo 4
tuple_4 = tuple_1 + tuple_2
tuple_5 = tuple_3 * 2
 
print(tuple_4)
print(tuple_5)
 
```
