Uma *função* é um bloco de código que executa uma tarefa específica quando a função é chamada (chamada). Você pode usar funções para tornar seu código reutilizável, melhor organizado e mais legível

```python
def your_function(optional_parameters):
    print(optional_parameters)
    return "fim da função"

your_function("parametro")
```

```python
def hello(name):    # define uma função
    print("Olá,", name)    # o corpo da função
 
 
name = input("Entre um valor: ")
 
hello(name) 
```

```python
# Parâmetros por palavra-chave

def introduction(first_name, last_name):
    print("Olá meu nome é", first_name, last_name)
 
introduction(first_name = "Fernanda", last_name = "Marques")
```

```python
def introduction(first_name="--", last_name="---"): # com valores padrão
    print("Olá meu nome é", first_name, last_name)

introduction(first_name="Nanda")
introduction()
```
