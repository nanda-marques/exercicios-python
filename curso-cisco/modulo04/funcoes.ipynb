Uma *função* é um bloco de código que executa uma tarefa específica quando a função é chamada (chamada). Você pode usar funções para tornar seu código reutilizável, melhor organizado e mais legível

```python
def your_function(optional_parameters):
    print(optional_parameters)
    return "fim da função"

your_function("parametro")
```

```python
def is_year_leap(year):
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

def days_in_month(year, month):
    if year < 1 or month < 1 or month > 12:
        return None
        
    february = 29 if is_year_leap(year) else 28
    month_lengths = [31, february, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    return month_lengths[month - 1]

def day_of_year(year, month, day):
    max_days_current_month = days_in_month(year, month)
    
    if max_days_current_month is None or day < 1 or day > max_days_current_month:
        return None

    total_days = 0
    for m in range(1, month):
        total_days += days_in_month(year, m)
        
    total_days += day
    
    return total_days

try:
    user_year = int(input("Digite o ano: "))
    user_month = int(input("Digite o mês (1-12): "))
    user_day = int(input("Digite o dia: "))

    resultado = day_of_year(user_year, user_month, user_day)

    if resultado is None:
        print("Data inválida! Verifique o ano, mês ou o limite de dias.")
    else:
        print(f"A data {user_day}/{user_month}/{user_year} corresponde ao dia {resultado} do ano.")

except ValueError:
    print("Por favor, digite apenas números inteiros válidos.")
```
