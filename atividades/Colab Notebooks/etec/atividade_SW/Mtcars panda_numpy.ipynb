```python
# import as bibliotecas e o arquivo MTcars.csv
import pandas as pd
import os
```

```python
caminho_arquivo = '/content/drive/MyDrive/ColabNotebooks/atividade_SW/Mtcars.csv'
mtcars = pd.read_csv(caminho_arquivo)
```

Monte uma tabela dos carros mais gastões de combustivel para os menos gastões

```python
#tabela maior consumo para menor consumo

carros_ordenados_consumo = mtcars.sort_values('mpg', ascending=True)
print("Carros ordenados do maior consumo (menos econômicos) para menor consumo (mais econômicos):")
print(carros_ordenados_consumo[['mpg']])
print("\n")
```

Podemos afirmar que os carros de maior cilindrada são os de maior consumo

```python
# relação entre cilindrada x consumo
correlacao_consumo_cilindrada = mtcars['mpg'].corr(mtcars['disp'])
print(f"Correlação entre cilindrada e consumo: {correlacao_consumo_cilindrada:.3f}")
if correlacao_consumo_cilindrada < -0.7:
    print("RESPOSTA: SIM - Há forte correlação negativa: carros de maior cilindrada são os de maior consumo")
elif correlacao_consumo_cilindrada < -0.3:
    print("RESPOSTA: Parcialmente - Há correlação negativa moderada")
else:
    print("RESPOSTA: NÃO - A correlação é fraca ou inexistente")
print("\n")
```

Os carros mais rápidos são motores em V ou em linha

```python
# forma de motores x tempo para 400m
tempo_vs = mtcars[mtcars['vs'] == 0]['qsec'].mean()  # Motores em V
tempo_reta = mtcars[mtcars['vs'] == 1]['qsec'].mean()  # Motores em linha
print(f"Tempo médio para 400m - Motores em V: {tempo_vs:.2f} segundos")
print(f"Tempo médio para 400m - Motores em linha: {tempo_reta:.2f} segundos")
if tempo_vs < tempo_reta:
    print("RESPOSTA: Motores em V são mais rápidos")
else:
    print("RESPOSTA: Motores em linha são mais rápidos")
print("\n")
```

O peso esta relacionado com o numero de cilindro

```python
# peso x numero de cilindro
correlacao_peso_cilindros = mtcars['wt'].corr(mtcars['cyl'])
print(f"Correlação entre peso e número de cilindros: {correlacao_peso_cilindros:.3f}")
if correlacao_peso_cilindros > 0.7:
    print("RESPOSTA: SIM - Há forte relação positiva")
elif correlacao_peso_cilindros > 0.3:
    print("RESPOSTA: Parcialmente - Há relação moderada")
else:
    print("RESPOSTA: NÃO - A relação é fraca")
print("\n")
```

Existe uma relação entre o numero de marchas em relação ao numero de carburadores

```python
# numero de marcha x nuimero de carburadores
relacao_gear_carb = mtcars.groupby('gear')['carb'].mean()
print("Relação entre número de marchas e número médio de carburadores:")
print(relacao_gear_carb)
print("RESPOSTA: Existe variação, mas geralmente mais marchas estão associadas a mais carburadores")
print("\n")
```

Mostre os 3 carros de maior consumo e os 3 de menor consumo

```python
# 3 de maior consumo
maior_consumo = mtcars.nlargest(3, 'mpg')[['mpg']]
print("3 carros de MAIOR consumo (menos econômicos):")
print(maior_consumo)
print("\n")
```

```python
# 3 de menor consumo
menor_consumo = mtcars.nsmallest(3, 'mpg')[['mpg']]
print("3 carros de MENOR consumo (mais econômicos):")
print(menor_consumo)
print("\n")
```
