```python
# Inicia um loop externo onde 'i' varia de 0 até 3 (representa as linhas)
for i in range(4):
    print("Novo I")
    # Inicia um loop interno onde 'j' varia de 0 até 3 (representa as colunas)
    for j in range(4):
        print("Novo J")
        # Verifica se o índice da coluna (j) ficou maior que o da linha (i)
        if j > i:
            # Se j for maior que i, interrompe imediatamente o loop interno atual
            break
        # Imprime o par ordenado (i, j) se o loop não tiver sido interrompido
        print((i, j))
```

```python
# Inicializa a variável 'x' com o valor 256
x = 256

# Inicializa o acumulador 'total' com o valor 0
total = 0

# Inicia um loop que continuará rodando enquanto 'x' for maior que 0
while x > 0:
    # Verifica se o 'total' acumulado até agora já ultrapassou o limite de 500
    if total > 500:
        # Se 'total' for maior que 500, interrompe o loop imediatamente
        break

    # Adiciona o valor atual de 'x' ao 'total' (mesmo que: total = total + x)
    total += x

    # Divide 'x' por 2 usando divisão inteira, descartando o resto (ex: 256 vira 128)
    x = x // 2

# Ao final do programa, as variáveis terminam com os seguintes valores:
# total = 504
# x = 4
```
