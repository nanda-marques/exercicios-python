```python
# P. 39 - ex: 6

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Dados das taxas de incremento anual
dados = [
    3.67, 1.82, 3.73, 4.10, 4.30,
    1.28, 8.14, 2.43, 4.17, 5.36,
    3.96, 6.54, 5.84, 7.35, 3.63,
    2.93, 2.82, 8.45, 5.28, 5.41,
    7.77, 4.65, 1.88, 2.12, 4.26,
    2.78, 5.54, 0.90, 5.09, 4.07,
]

# Configuração da figura
# plt.figure(figsize=(10, 3))

# Construção do gráfico de dispersão 1D com um leve jitter para evitar sobreposição
sns.stripplot(
    x=dados,
    size=8,
    color="skyblue",
    edgecolor="black",
    linewidth=1,
    # jitter=0.05,
)

# Personalização
plt.title(
    "Gráfico de Dispersão Unidimensional (1D Scatter Plot)",
    fontsize=12,
    fontweight="bold",
    pad=15, # distância do título
)
plt.xlabel(
    "Taxa Média Geométrica de Incremento Anual (por 100 hab.)",
    fontsize=10
)
plt.yticks([])  # Remove o eixo Y por ser unidimensional
plt.grid(axis="x", linestyle="--")
plt.tight_layout()

# Exibir o gráfico
plt.show()
```
