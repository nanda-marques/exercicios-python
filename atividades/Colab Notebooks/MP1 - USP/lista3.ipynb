```python
massa = float(input())
num_voltas = int(input())
maior_aceleracao = 0
volta_mais_rapida = 0
newtons = 0

for volta in range(1, num_voltas + 1):
  newtons = float(input())
  aceleracao = newtons / massa
  if aceleracao > maior_aceleracao:
    maior_aceleracao = aceleracao
    volta_mais_rapida = volta

print('melhor_volta:', volta_mais_rapida)
print('maior aceleração:', f"{maior_aceleracao:.1f}")
```

o sistema eletrônico computa quantos senadores votaram a favor (F) ou contra (C) a matéria em questão. Os senadores também podem se de ausentar da votação (A), votar em branco (B) ou nulo (N)

```python
num_senadores = int(input())

a_favor = 0
contra = 0
outros = 0

for i in range(num_senadores):
    voto = input()
    if voto == 'F':
        a_favor += 1
    elif voto == 'C':
        contra += 1
    else:
        outros += 1

print(f"a favor: {a_favor}")
print(f"contra: {contra}")
print(f"outros: {outros}")
```

verificar se o tamanho da string resultante é o esperado para um CPF (11 caracteres)

verificar se todos os caracteres são um dos algarismos entre '0' e '9'

Imprima

o CPF sem pontos e traços, se estiver tudo certo

ENCODING ERROR se houver algum caractere inválido

SIZE ERROR se o tamanho não for válido

```python
cpf = input("Digite o CPF: ")

cpf_limpo = cpf.replace(".", "").replace("-", "")
if len(cpf_limpo) != 11 and not cpf_limpo.isdigit():
    print("ENCODING ERROR")
    print("SIZE ERROR")
elif not cpf_limpo.isdigit():
    print("ENCODING ERROR")
elif len(cpf_limpo) != 11:
    print("SIZE ERROR")
else:
    print(cpf_limpo)
```

```python
tamanho = int(input())
pin = []
for i in range(tamanho):
  algarismo = input()
  pin.append(algarismo)
if len(pin) == len(set(pin)):
  print("OK")
else:
  print("ERRO")
```

- mais vogais que dígitos

- pelo menos dois dígitos

- pelo menos 8 caracteres

```python
senha = input()

vogais = 0
digitos = 0

vogais_lista = "aeiouAEIOU"

for i in senha:
    if i in vogais_lista:
        vogais += 1
    elif i.isdigit():
        digitos += 1

if len(senha) >= 8 and digitos >= 2 and vogais > digitos:
    print("OK")
else:
    print("ERRO")
```
