# Problema 7
float()

Um valor do tipo ponto flutuante, obtido com a função float(), é uma aproximação de um valor real feita nos nossos computadores.

Linha 1

   A função input() obtém uma sequência de caracteres do dispositivo de entrada

   A função float() converte a sequência lida para um valor que pode ser interpretado por nosso programa como um número com casas decimais!

Linhas 2 a 6: impressão à moda antiga

   Nessas linhas utilizamos print() para apresentar o conteúdo apontado por valor com uma, duas, três, quatro e cinco casas decimais

   A sequência '%.nf' especifica que o valor deve ser apresentado com n casas decimais

   Note que, nesses casos, o símbolo % é utilizado como separador entre sequência  '%.nf' e o nome da variável valor

```python
valor = float(input())
print('%.2f' % valor)
print('%.4f' % valor)
print('%.6f' % valor)
print('%.8f' % valor)
print('%.10f' % valor)
```

# Problema 21
Uma boa noite de sono consiste em 5 ou 6 ciclos completos de sono, sendo que cada ciclo tem, em média, cerca de 90 minutos.

- Dado o numero de horas e minutos que alguém dormiu, seu programa deve calcular o número de ciclos de 90 minutos correspondente ao número total de minutos que alguém dormiu.

```python
horaSono = int(input())
minSono = int(input())

total_minutos = horaSono * 60 + minSono
ciclo = total_minutos // 90

if 5 <= ciclo <= 6:
  print(f'{ciclo} ciclos indica uma boa noite de sono')
else:
  print(f'{ciclo} ciclos não indica uma boa noite de sono')
```
