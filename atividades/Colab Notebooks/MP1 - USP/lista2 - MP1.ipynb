#Problema 2
Seu programa deve ler três linhas. Na primeira, um símbolo correspondendo a uma operação entre dois inteiros. Nas linhas seguintes, um valor inteiro cada. O programa deve calcular o resultado da operação entre os dois inteiros e imprimir o resultado.

```python
simbolo = input()
num1 = int(input())
num2 = int(input())

if simbolo == '+':
  print(num1 + num2)
elif simbolo == '-':
  print(num1 - num2)
elif simbolo == '*':
  print(num1 * num2)
elif simbolo == '/':
  print(num1 / num2)
elif simbolo == '//':
  print(num1 // num2)
elif simbolo == '%':
  print(num1 % num2)
elif simbolo == '**':
  print(num1 ** num2)
```

# Problema 15

# Problema 20
Tem o que precisa na senha?

Dado um caractere qual sua classificacao?

- vogal (minúscula ou maiúscula), algarismo, carectere especial ou outro

- um caractere especial é um entre os seguintes: @#$%&*()_-+=!


```python
caractere = input()
if caractere in 'aeiouAEIOU':
  print('vogal')
elif caractere in '1234567890':
  print('algarismo')
elif caractere in '@#$%&*()_-+=!':
  print('especial')
else:
  print('outro')
```
