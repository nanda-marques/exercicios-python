Dados os dados de uma mulher, concluinte de graduação ou pós-graduação, cujo salário se sabe ser desigual em relação aos seus pares do gênero masculino na proporção reportada acima, calcule quanto deveria ser salário se ela ganhasse igual a um homem que exerce a mesma função.
Seu programa lê duas linhas: a primeira informa se a profissional possui graduação (G) ou pós-graduação (PG); a segunda informa seu salario atual com um valor inteiro. Como saída, o programa imprime o salário que a mulher deveria receber se não houvesse a desigualdade identificada acima.
Lembrete: para imprimir o valor de xis com duas casas decimais, podemos utilizar print('%.2f' % xis)
homens recebem 19% a mais do que as mulheres para exercer a mesma função.
Segundo o levantamento, as mulheres representam 56% dos profissionais com pós-graduação, especialização ou MBA e 52% em formação superior. Mas mesmo assim ganham, respectivamente, 43% a 38% a menos em relação aos homens com o mesmo grau de escolaridade.

Em seu estágio, sua supervisora pediu para você criar um programa que processa os dados de uma pesquisa sobre a diferença salarial entre homens e mulheres em uma empresa, sabendo que a empresa tem pelo menos um funcionário que se identifica com cada gênero.

Os dados da pesquisa foram armazenados de forma que os salários dos homens são informados como valores inteiros positivos e o das mulheres como valores inteiros negativos. O fim da entrada de dados é indicado pelo valor nulo.

Seu programa deve imprimir o salário médio de mulheres e homens, nessa ordem.

```python
salario_homem = 0
salario_mulher = 0
total_homens = 0
total_mulheres = 0

while True:
    salario = int(input())
    if salario == 0:
        break
    if salario > 0:
        salario_homem += salario
        total_homens += 1
    elif salario < 0:
        salario_mulher += -salario
        total_mulheres += 1

if total_homens > 0:
    media_homens = salario_homem / total_homens

if total_mulheres > 0:
    media_mulheres = salario_mulher / total_mulheres

print('%.2f' % media_mulheres)
print('%.2f' % media_homens)
```

Nosso grupo tem uma meta a alcançar, e cada menina traz sua energia para ajudar.



Neste desafio, você deve somar as energias das meninas, representadas por algarismos, até que juntas alcancem ou ultrapassem um valor-alvo fornecido no início.



No final, reporte a soma total e registre como cada contribuição ajudou a formar essa força coletiva.

```python
valorAlvo = int(input())
soma = 0
energiaTotal = []

while soma < valorAlvo:
    sequencia = input()
    for energia in sequencia:
        energiaTotal.append(energia)
        soma = int(energia) + soma
        if soma >= valorAlvo:
            break
    if soma >= valorAlvo:
        break
print(soma, energiaTotal)
```

Quando a Joaninha é ligada, o programa captura da bateria um valor inteiro entre 0 e 100, correspondente à porcentagem de carga disponível

A Joaninha só funciona com carga mínima maior que 5%

Antes de cada movimento, o programa obtém informações de dois sensores. O primeiro sensor informa B de bateu ou L se a área livre à frente. O segundo sensor informa A se tem um Abismo, ou P se há Piso para avançar

Para controlar a Joaninha, seu programa emite os comandos virar, em caso de necessidade, ou avançar, sempre que possível. Quando não há bateria disponível para trabalhar, seu programa avisa que é hora de recarregar.  Junto com cada comando, seu sempre programa informa o quanto de bateria resta na Joaninha.

Cada movimento de avançar consome 1% de bateria. Cada movimento de virar consome  5% de bateria.

Entrada

Primeira linha: um valor inteiro entre 0 e 100

A seguir, um número indeterminado de pares de linhas com uma letra cada

- o primeiro valor do par contém B ou L

- o segundo valor do par contém A ou P

```python
carga = int(input())

while carga > 5:
    comando1 = input().strip()
    comando2 = input().strip()

    if comando1 == 'B':
        carga -= 5
        print(f"virar: {carga}")
    elif comando1 == 'L' and comando2 == 'P':
        carga -= 1
        print(f"avançar: {carga}")
    elif comando1 == 'L' and comando2 == 'A':
        carga -= 5
        print(f"virar: {carga}")

if carga <= 5:
    print(f"recarregar: {carga}")
```
