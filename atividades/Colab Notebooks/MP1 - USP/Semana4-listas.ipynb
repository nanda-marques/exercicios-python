#**Semana 4 (Python) Listas**

Uma *estrutura de dados* é uma forma de organizar e armazenar dados de maneira eficiente, permitindo o acesso e a manipulação desses dados de acordo com as necessidades de nosso programa

*Listas* em Python são *estruturas de dados* fundamentais e extremamente versáteis em muitos aspectos, inclusive:
- Armazenamento de dados
- Acesso a elementos
- Iteração
- Inclusão e remoção de elementos
- Ordenação
- Flexibilidade
---

##Armazenamento de Dados

Listas permitem armazenar coleções de itens em uma única variável
- isso é útil quando precisamos trabalhar com conjuntos de dados relacionados

```
# Lista com números inteiros
numeros = [1, 2, 3, 4, 5]

# Lista com nomes de pessoas
nomes = ['Bia', 'Duda', 'Valéria', 'Olga', 'Cris']
```

---
##Acesso a Elementos
Listas permitem nosso acesso aos seus elementos individuais por meio de índices

```
alunas = ['Bia', 'Duda', 'Valéria', 'Olga', 'Cris']
primeira_aluna = alunas[0]
última_aluna = alunas[-1]
```

---
## Iteração
Listas permitem que possamos iterar sobre cada um de seus elementos

```
alunas = ['Bia', 'Duda', 'Valéria', 'Olga', 'Cris']
for aluna in alunas:
  print(aluna)
```

````python
alunas = ['Bia', 'Duda', 'Valéria', 'Olga', 'Cris']
for aluna in alunas:
  print(aluna)
````

---
#Inclusão e remoção de elementos

Listas podem ser modificadas dinamicamente: permitem a inclusão e remoção de elementos

```
alunas = ['Bia', 'Duda', 'Valéria', 'Olga', 'Cris']
aluna_primeira = alunas.pop(0)
aluna_ultima = alunas.pop()
aluna_nova = 'Ana'
alunas.append(aluna_nova)
alunas.remove(aluna_nova)
```

````python
alunas = ['Bia', 'Duda', 'Valéria', 'Olga', 'Cris']
print(alunas)
aluna_primeira = alunas.pop(0) # remove a primeira aluna da lista e deixa apontado por aluna_primeira
aluna_ultima = alunas.pop()   # remove a última aluna da lista e deixa apontado por aluna_ultima
aluna_nova = 'Ana'
print(aluna_primeira, aluna_ultima, aluna_nova)
print(alunas)
alunas.append(aluna_nova)     # insere aluna_nova no final da lista
print(alunas)
alunas.remove(aluna_nova)     # remove aluna_nova no final da lista
print(alunas)
alunas.insert(0, aluna_nova)  # insere aluna_nova no início da lista
print(alunas)
````

---
##Ordenação

Listas oferecem métodos convenientes para ordenarmos seus elementos

```
alunas = ['Bia', 'Duda', 'Valéria', 'Olga', 'Cris']
alunas.sort()
```

````python
alunas = ['Bia', 'Duda', 'Valéria', 'Olga', 'Cris']
print(alunas)
alunas.sort()
print(alunas)
````

---
## Flexibilidade

Listas podem conter diferentes tipos de dados (números, strings, outras listas, etc.)

```
alunas_reprovadas = []
professora = ['Maria da Graça', 62, ['Graduação', 'Mestrado', 'Doutorado', 'Livre-docência', 'Titular']]

```

````python
alunas_reprovadas = []
professora = ['Maria da Graça', 62, ['Graduação', 'Mestrado', 'Doutorado', 'Livre-docência', 'Titular']]
print(alunas_reprovadas)
print(professora)
````

###**Problema: Lista de compras**
Monte uma lista de compras com os itens dados um por linha; uma linha com ponto final indica fim da lista de compras. Imprima a lista original e a lista ordenada.


````python
lista_de_compras = []
while True:
  item = input()
  if item == '.':
    break
  else:
    lista_de_compras.append(item)
print(lista_de_compras)
lista_de_compras.sort()
print(lista_de_compras)
````
