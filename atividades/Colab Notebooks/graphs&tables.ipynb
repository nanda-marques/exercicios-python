```python
import matplotlib.pyplot as plt
import numpy as np

# Dados (ajuste com os valores oficiais se quiser)
trimestres = [
    '2022T1', '2022T2', '2022T3', '2022T4',
    '2023T1', '2023T2', '2023T3', '2023T4',
    '2024T1', '2024T2', '2024T3', '2024T4',
    '2025T1'
]

taxa_desemprego = [
    11.1, 9.8, 8.7, 7.9,
    8.8, 8.3, 7.7, 7.4,
    8.2, 7.9, 7.7, 7.6,
    7.9
]

# Configurando o gráfico de barras
x = np.arange(len(trimestres))

plt.figure(figsize=(12, 6))
bars = plt.bar(x, taxa_desemprego, color='skyblue', edgecolor='black')

# Adicionando os valores em cima de cada barra
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 0.2,
             f'{height:.1f}%', ha='center', fontsize=9)

plt.title('Taxa de Desemprego no Brasil por Trimestre (2022 a 2025)', fontsize=14)
plt.xlabel('Trimestre', fontsize=12)
plt.ylabel('Taxa de Desemprego (%)', fontsize=12)
plt.xticks(x, trimestres, rotation=45)
plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Dados (você pode ajustar com os números exatos se quiser)
trimestres = [
    '2022T1', '2022T2', '2022T3', '2022T4',
    '2023T1', '2023T2', '2023T3', '2023T4',
    '2024T1', '2024T2', '2024T3', '2024T4',
    '2025T1'
]

taxa_desemprego = [
    11.1, 9.8, 8.7, 7.9,
    8.8, 8.3, 7.7, 7.4,
    8.2, 7.9, 7.7, 7.6,
    7.9
]

x = np.arange(len(trimestres))

plt.figure(figsize=(12, 7))

# Linha principal
plt.plot(x, taxa_desemprego, color='black', linewidth=2.5, marker='o', markersize=8, markerfacecolor='white', markeredgewidth=2)

# Adicionando os valores só nos pontos inicial e final
plt.text(x[0], taxa_desemprego[0] + 0.3, f'{taxa_desemprego[0]:.1f}%', ha='center', fontsize=10, fontweight='bold')
plt.text(x[-1], taxa_desemprego[-1] + 0.3, f'{taxa_desemprego[-1]:.1f}%', ha='center', fontsize=10, fontweight='bold')

# Configurações do gráfico
plt.title('Taxa de Desemprego no Brasil por Trimestre (2022 a 2025)', fontsize=16, fontweight='bold')
plt.xlabel('Trimestres', fontsize=12)
plt.ylabel('Taxa de Desemprego (%)', fontsize=12)
plt.xticks(x, trimestres, rotation=45, fontsize=10)
plt.yticks(fontsize=10)
plt.grid(axis='y', linestyle='--', color='gray', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Criando o DataFrame
dados = {
    'Condição': [
        'População em idade de trabalhar',
        'Força de trabalho (PEA)',
        'Ocupados',
        'Desocupados',
        'Fora da força de trabalho'
    ],
    'Quantidade (mil pessoas)': [
        178_532,
        108_340,
        99_777,
        8_563,
        70_192
    ]
}

df = pd.DataFrame(dados)

# Criando a figura
fig, ax = plt.subplots(figsize=(8, 3))  # Ajuste o tamanho conforme quiser

# Removendo eixos
ax.axis('off')

# Criando a tabela na imagem
tabela = ax.table(
    cellText=df.values,
    colLabels=df.columns,
    cellLoc='left',
    loc='center',
    colLoc='left'
)

# Melhorando o layout
tabela.scale(1, 2)  # Largura x Altura (ajuste se quiser mais espaço entre as linhas)

# Bordas pretas
for key, cell in tabela.get_celld().items():
    cell.set_linewidth(1)
    cell.set_edgecolor('black')

# Salvando como imagem PNG
plt.savefig('tabela_desemprego.png', dpi=300, bbox_inches='tight')
plt.show()
```
