```python
class Retangulo:

 def __init__(self, lado_a, lado_b):
    self.a = lado_a
    self.b = lado_b

 def muda_valor(self, novo_a, novo_b):
    self.a = novo_a
    self.b = novo_b

 def retorna_lado(self):
    print(f'O Retângulo possui dimensões {self.a} m {self.b} m')

 def area(self):
    return self.a * self.b

import math

while True:
    try:
        piso_a_input = input("Digite um lado do piso: ")
        piso_a = float(piso_a_input)

        piso_b_input = input("Digite o outro lado do piso: ")
        piso_b = float(piso_b_input)

    except ValueError:
        print("Entrada inválida. Por favor, digite um número.")
        continue

    piso = Retangulo(piso_a, piso_b)

    try:
        az_a_input = input("Digite o lado do azulejo: ")
        az_a = float(az_a_input)

        az_b_input = input("Digite o outro lado do azulejo: ")
        az_b = float(az_b_input)

    except ValueError:
        print("Entrada inválida. Por favor, digite um número.")
        continue

    Azulejo = Retangulo(az_a, az_b)

    area_piso = piso.area()
    area_az = Azulejo.area()

    if area_az == 0: # Prevent division by zero if az_a or az_b was 0
        print("A área do azulejo não pode ser zero. Por favor, insira dimensões válidas.")
        continue

    qntd_az = area_piso / area_az

    if area_piso % area_az == 0:
        print(f'A quantidade exata de azulejos para preencher o piso é de {qntd_az}')

    else:
        print(f'A quantidade mínima de azulejos para preencher o piso é de {math.ceil(qntd_az)}')
    break
```
