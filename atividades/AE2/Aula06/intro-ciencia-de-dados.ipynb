```python
# deletando duplicatas
duplicados = dados[dados.duplicated(keep='first')]
print(duplicados)

dados.drop_duplicates(keep='first', inplace=True)
```

```python
# NaN - Not a Number
# Pode decodificar um valor faltante, mas é definido como float

dados = dados.replace(['nao_sei'], np.nan)

dados['key'] = dados['key'].replace([0.177], np.nan)

dados['audio_mode'] = dados['audio_mode'].replace(['0.105'], np.nan)

dados['speechiness'] = dados ['speechiness'].replace(['0.nao_sei'], np.nan)

dados['time_signature'] = dados['time_signature'].replace(['0.7', '2800000000'], np.nan)
```
