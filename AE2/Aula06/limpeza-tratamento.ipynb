```python
import streamlit as st
```

```python
# entrada de dados

nome = st.text_input("Digite seu nome: ")
idade = st.number_input("Idade: ", min_value=0)
opcao = st.selectbox("Escolha um produto: ", ["A", "B", "C"])
```

```python
# Exibição de dados

st.dataframe(df) # tabela interativa
st.table(df.head()) # tabela estática
```

```python
# Gráficos

st.line_chart(df["vendas"]) # linhas
st.bar_chart(df["quantidade"]) # barras
```

```python
# Upload de arquivos

arquivo = st.file_uploader("Envie um CSV: ")

if arquivo:
    df = pd.read_csv(arquivo)
    st.dataframe(df)
```

```python
filtro = st.slider("Filtrar vendas acima de: ", 0, 1000, 100)

df_filtrado = df[df["vendas"] > filtro]

st.dataframe(df_filtrado)
```

```python
if st.button("Clique aqui!"):
    st.write("Botão pressionado.")
```

```python
opcoes = st.multiselect("Escolha frutas: ", "Maçã", "Banana", "Laranja")
st.write("Selecionadas: ", opcoes)
```
